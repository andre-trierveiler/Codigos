```python

 
Select  = f"""
            Select * from dl_Gestao_recursos.vw_fra_vertice_fixo
            where dt> '2001-01-01'
            and fra_du in ('21x126','126x252','252x315')
            """
 
df = pd.read_sql(Select, session) 
df.DT = pd.to_datetime(df.DT, format='%Y-%m-%d') 
pd.set_option('mode.chained_assignment', None)

df_back = df
```

```python
df = df_back
df = df.sort_values(by=['FRA', 'DT'])
df['DIF_FRA_5'] =  df['TX_FRA'] - df.groupby('FRA')['TX_FRA'].shift(5)
df['DIF_FRA_10'] = df['TX_FRA'] - df.groupby('FRA')['TX_FRA'].shift(10) 
df['DIF_FRA_21'] = df['TX_FRA'] - df.groupby('FRA')['TX_FRA'].shift(21) 
df['DIF_FRA_63'] = df['TX_FRA'] - df.groupby('FRA')['TX_FRA'].shift(62)

df['DIF_SPREAD_5'] = df['TX_SPREAD'] - df.groupby('FRA')['TX_SPREAD'].shift(5)
df['DIF_SPREAD_10'] = df['TX_SPREAD'] - df.groupby('FRA')['TX_SPREAD'].shift(10) 
df['DIF_SPREAD_21'] = df['TX_SPREAD'] - df.groupby('FRA')['TX_SPREAD'].shift(21)
df['DIF_SPREAD_63'] = df['TX_SPREAD'] - df.groupby('FRA')['TX_SPREAD'].shift(62)

df['DIF_TX_5'] = df['TAXA_A'] - df.groupby('FRA')['TAXA_A'].shift(5)
df['DIF_TX_10'] = df['TAXA_A'] - df.groupby('FRA')['TAXA_A'].shift(10) 
df['DIF_TX_21'] = df['TAXA_A'] - df.groupby('FRA')['TAXA_A'].shift(21)
df['DIF_TX_63'] = df['TAXA_A'] - df.groupby('FRA')['TAXA_A'].shift(62)


plot.rcParams['figure.figsize'] = (10,5)
sns.set(style="darkgrid", palette="Set2")
f = plot.figure(1)
sns.lineplot(x='DT', y='DIF_FRA_63', data=df, hue='FRA_DU').set(title='Histórico Variação de 5 Dias')
```
