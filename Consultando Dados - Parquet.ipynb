```python
path = r"//nas03/gestao_recursos/Dados/dataset/"
import pandas as pd

idka = pd.read_parquet(path+'anbima/indices/idka/')

imas = pd.read_parquet(path+'anbima/indices/ima/')

idas = pd.read_parquet(path+'anbima/indices/ida/')
```

Para estes índices anbima, pode-se fazer a análise que queira.

Por exemplo, vamos fazer uma análise rápida e comparativa entre CDI e e IDkA Pré 2y

```python
cdi = pd.read_parquet(path+'indices/CDI').set_index('data_referencia')['value'].pct_change()
cdi.index=pd.to_datetime(cdi.index)

idka_pre_2y = idka.set_index('data_referencia').query("nome=='IDkA Pré 2A'")['numero_indice'].pct_change()
idka_pre_2y.index=pd.to_datetime(idka_pre_2y.index)

imasb5 = imas.set_index('data_referencia').query("indice=='IMA-B 5'")['numero_indice'].pct_change()
imasb5.index=pd.to_datetime(imasb5.index)
```

```python
df=pd.concat([cdi,idka_pre_2y,imasb5],axis=1).dropna()
df.columns=['CDI','IDkA Pré 2y','IMA-B 5']
```

```python
import quantstats as qs
```

```python
qs.reports.html(returns = df['IDkA Pré 2y'], benchmark =df['CDI'] ,title='Strategy =  IDkA Pré 2y')
```
