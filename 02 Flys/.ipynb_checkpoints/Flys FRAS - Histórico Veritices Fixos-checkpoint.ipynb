```python
def FLY(lista_fly, dt_inicial = '01/01/2004', dt_final = cal.offset(datetime.now(),-1).strftime('%d/%m/%Y'), span = False, fim_mes = True):

    
    start = time.time() 
    lista_string = "','".join(map(str,lista_fly))
    data_inicial = datetime.strptime(dt_inicial, "%d/%m/%Y").date()
    data_final = datetime.strptime(dt_final, "%d/%m/%Y").date()
    dt_final_sql = data_final.strftime('%Y-%m-%d')
    dt_inicial_sql = data_inicial.strftime('%Y-%m-%d')
    if fim_mes  == True:
       fl_fim_mes_sql = 'DU.FL_FIM_MES'
    else:
        fl_fim_mes_sql = '1'
        
    Select = f"""
    SELECT DT
    , FLY
    , FLY_ABC AS TAXA
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_SUP_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_INF_6M

    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) AS MEDIA_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_SUP_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_INF_21
    
    , MIN(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MINIMO_6M
    , MAX(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MAXIMO_6M
    , STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS DEVPAD_AM_6M
    
    , FWD_AB
    , SPREAD_AB
    , CDI
    , FL_FIM_MES
    
    FROM 
    (SELECT DI_B.DATA as DT
    , to_char(ROUND(CAST(DI_A.VERTICE AS FLOAT)/252,2))  || 'x' || TO_CHAR(ROUND(CAST(DI_B.VERTICE AS FLOAT)/252,2)) || 'x' || TO_CHAR(ROUND(CAST(DI_C.VERTICE AS FLOAT)/252,2)) AS FLY
    , DI_A.VERTICE AS CONTRATO_A, DI_B.VERTICE AS CONTRATO_B, DI_C.VERTICE AS CONTRATO_C
    , DI_A.TAXA AS TAXA_A, DI_B.TAXA AS TAXA_B, DI_C.TAXA AS TAXA_C
    
    , ((1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)))-1 AS FWD_AB
    , ((1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)))-1 AS FWD_BC
    ,  (1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)) - (1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)) AS FLY_ABC
    ,  (DI_B.TAXA - DI_A.TAXA) as SPREAD_AB
    ,  (DI_C.TAXA - DI_B.TAXA) as SPREAD_BC
    ,  (DI_C.TAXA - DI_B.TAXA) - (DI_B.TAXA - DI_A.TAXA) as FLY_PONTOS
    
    , (CAST(DI_B.VERTICE AS FLOAT)/(1+DI_B.TAXA) - CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA))/(CAST(DI_A.VERTICE AS FLOAT)/(1+DI_A.TAXA)-CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA)) AS ALFA
    , ((IDX.VALOR/3000+1)**252.000000-1)*100 AS CDI
    , NVL(DU.FL_FIM_MES, DU.FL_D1) as FL_FIM_MES
    FROM DL_GESTAO_RECURSOS.CURVA DI_B
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_A ON DI_A.DATA = DI_B.DATA AND DI_A.VERTICE < DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_C ON DI_C.DATA = DI_B.DATA AND DI_C.VERTICE > DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.IDX_INDEX IDX ON IDX.DATA = DI_B.DATA AND IDX.CODINDEX = 'CDI'
    INNER JOIN dl_Gestao_recursos.vw_du DU ON DU.DT = DI_B.DATA AND ({fl_fim_mes_sql} = 1 or du.fl_d1 = 1)
    --WHERE 
    --DI_B.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_A.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_C.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520')
    ) VW
    WHERE VW.DT between '{data_inicial}' and '{data_final}'
    AND VW.FLY IN ('{lista_string}')
    """

   
    fly_df = pd.read_sql(Select, session) 
    fly_df.DT = pd.to_datetime(fly_df.DT, format='%Y-%m-%d')
    if fim_mes  == True:
        fly_df = fly_df[fly_df.FL_FIM_MES == 1]
   
    for x in lista_fly:   
        fly_grafico = fly_df[fly_df.FLY == x] 
        fly_grafico = fly_grafico.sort_values(by=['DT'])
        fly_grafico = fly_grafico.reset_index() 
        fly_grafico = fly_grafico[fly_grafico.FL_FIM_MES == 1]
        pd.set_option('mode.chained_assignment', None)
    
        #fig = make_subplots(specs=[[{"secondary_y": True}]])
        #fig.add_vline(x=datetime(2014,11,15), line_width=3, line_dash="dash", line_color="green")
        TAXA = go.Scatter(x = fly_grafico.DT,y = fly_grafico['TAXA']*10000,mode = 'lines', name = 'Taxa', line = {'color': 'dodgerblue', 'width':1.5}, yaxis = 'y1')
        MEDIA = go.Scatter(x = fly_grafico.DT,y = fly_grafico['MEDIA_6M']*10000,mode = 'lines', name = 'Media', line = {'color': 'violet', 'width':1.5},yaxis = 'y1')
        LIM_INF = go.Scatter(x = fly_grafico.DT,y = fly_grafico['LIM_INF_6M']*10000,mode = 'lines', name = 'Liminte Inferior', line = {'color': 'violet','width':1},yaxis = 'y1')
        LIM_SUP = go.Scatter(x = fly_grafico.DT,y = fly_grafico['LIM_SUP_6M']*10000,mode = 'lines', name = 'Limite Superior', line = {'color': 'violet','width':1},yaxis = 'y1')
    
        CDI = go.Scatter(x = fly_grafico.DT,y = fly_grafico['CDI'],mode = 'lines', name = 'CDI', line = {'color': 'goldenrod', 'width':0.5,'dash':'dash'}, yaxis = 'y2')
      
        data  = [TAXA, MEDIA, LIM_SUP,LIM_INF, CDI]
        layout = dict(title = x,
                           xaxis = dict(title = 'Data')
                           ,yaxis = dict(title = 'Taxa')
                           #,xaxis_rangeslider_visible=True
                           ,yaxis2=dict(title='CDI',
                                       overlaying='y1',
                                       side='right')
                           )

        fig = dict(data=data, layout=layout)
        pio.show(fig)  
        end = time.time()
        print("Execution Time: %s secs." % "{:.2f}".format(end-start))
```

```python
def FLY_TESTE(lista_fly, dt_inicial = '01/01/2004', dt_final = cal.offset(datetime.now(),-1).strftime('%d/%m/%Y'), span = False, fim_mes = True):

    
    start = time.time() 
    lista_string = "','".join(map(str,lista_fly))
    data_inicial = datetime.strptime(dt_inicial, "%d/%m/%Y").date()
    data_final = datetime.strptime(dt_final, "%d/%m/%Y").date()
    dt_final_sql = data_final.strftime('%Y-%m-%d')
    dt_inicial_sql = data_inicial.strftime('%Y-%m-%d')
    if fim_mes  == True:
       fl_fim_mes_sql = 'DU.FL_FIM_MES'
    else:
        fl_fim_mes_sql = '1'
        
    Select = f"""
    SELECT DT
    , FLY
    , FLY_ABC AS TAXA
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_SUP_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_INF_6M

    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) AS MEDIA_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_SUP_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_INF_21
    
    , MIN(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MINIMO_6M
    , MAX(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MAXIMO_6M
    , STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS DEVPAD_AM_6M
    
    , FWD_AB
    , SPREAD_AB
    , CDI
    , FL_FIM_MES
    
    FROM 
    (SELECT DI_B.DATA as DT
    , to_char(ROUND(CAST(DI_A.VERTICE AS FLOAT)/252,2))  || 'x' || TO_CHAR(ROUND(CAST(DI_B.VERTICE AS FLOAT)/252,2)) || 'x' || TO_CHAR(ROUND(CAST(DI_C.VERTICE AS FLOAT)/252,2)) AS FLY
    , DI_A.VERTICE AS CONTRATO_A, DI_B.VERTICE AS CONTRATO_B, DI_C.VERTICE AS CONTRATO_C
    , DI_A.TAXA AS TAXA_A, DI_B.TAXA AS TAXA_B, DI_C.TAXA AS TAXA_C
    
    , ((1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)))-1 AS FWD_AB
    , ((1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)))-1 AS FWD_BC
    ,  (1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)) - (1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)) AS FLY_ABC
    ,  (DI_B.TAXA - DI_A.TAXA) as SPREAD_AB
    ,  (DI_C.TAXA - DI_B.TAXA) as SPREAD_BC
    ,  (DI_C.TAXA - DI_B.TAXA) - (DI_B.TAXA - DI_A.TAXA) as FLY_PONTOS
    
    , (CAST(DI_B.VERTICE AS FLOAT)/(1+DI_B.TAXA) - CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA))/(CAST(DI_A.VERTICE AS FLOAT)/(1+DI_A.TAXA)-CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA)) AS ALFA
    , ((IDX.VALOR/3000+1)**252.000000-1)*100 AS CDI
    , NVL(DU.FL_FIM_MES, DU.FL_D1) as FL_FIM_MES
    FROM DL_GESTAO_RECURSOS.CURVA DI_B
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_A ON DI_A.DATA = DI_B.DATA AND DI_A.VERTICE < DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_C ON DI_C.DATA = DI_B.DATA AND DI_C.VERTICE > DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.IDX_INDEX IDX ON IDX.DATA = DI_B.DATA AND IDX.CODINDEX = 'CDI'
    INNER JOIN dl_Gestao_recursos.vw_du DU ON DU.DT = DI_B.DATA AND ({fl_fim_mes_sql} = 1 or du.fl_d1 = 1)
    --WHERE 
    --DI_B.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_A.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_C.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520')
    ) VW
    WHERE VW.DT between '{data_inicial}' and '{data_final}'
    AND VW.FLY IN ('{lista_string}')
    """

   
    fly_df = pd.read_sql(Select, session) 
    fly_df.DT = pd.to_datetime(fly_df.DT, format='%Y-%m-%d')
    if fim_mes  == True:
        fly_df = fly_df[fly_df.FL_FIM_MES == 1]
   

    valores_fly = fly_df['FLY'].unique()
    
    # Criar a figura interativa
    fig = px.line(fly_df, x='DT', y='TAXA', color='FLY')
    
    # Adicionar o eixo secundário e plotar a coluna 'Taxa_b'
    fig.add_trace(px.line(fly_df, x='DT', y='MEDIA_6M', color='FLY').data[0])
    fig.update_traces(yaxis='y2')
    
    # Adicionar o filtro interativo
    fig.update_layout(
        updatemenus=[
            dict(
                buttons=[
                    dict(
                        label=FLY,
                        method="update",
                        args=[{"visible": [FLY == f for f in fly_df['FLY']]}],
                    )
                    for FLY in valores_fly
                ],
                direction="down",
                pad={"r": 10, "t": 10},
                showactive=True,
                x=0.1,
                xanchor="left",
                y=1.1,
                yanchor="top"
            ),
        ]
    )
    
    # Configurar o eixo secundário
    fig.update_layout(
        yaxis2=dict(
            title='CDI',
            overlaying='y',
            side='right'
        )
    )
    
    # Exibir o gráfico
    fig.show()
    print("Execution Time: %s secs." % "{:.2f}".format(end-start))
        
  
```

```python
def FLY(lista_fly, dt_inicial = '01/01/2004', dt_final = cal.offset(datetime.now(),-1).strftime('%d/%m/%Y'), span = False, fim_mes = True):

    
    start = time.time() 
    lista_string = "','".join(map(str,lista_fly))
    data_inicial = datetime.strptime(dt_inicial, "%d/%m/%Y").date()
    data_final = datetime.strptime(dt_final, "%d/%m/%Y").date()
    dt_final_sql = data_final.strftime('%Y-%m-%d')
    dt_inicial_sql = data_inicial.strftime('%Y-%m-%d')
    if fim_mes  == True:
       fl_fim_mes_sql = 'DU.FL_FIM_MES'
    else:
        fl_fim_mes_sql = '1'
        
    Select = f"""
    SELECT DT
    , FLY
    , FLY_ABC AS TAXA
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_SUP_6M
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as LIM_INF_6M

    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) AS MEDIA_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) + 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_SUP_21
    , AVG(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW) - 2*(STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 21 PRECEDING AND CURRENT ROW)) as LIM_INF_21
    
    , MIN(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MINIMO_6M
    , MAX(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MAXIMO_6M
    , STDDEV_SAMP(FLY_ABC) OVER (PARTITION BY FLY ORDER BY DT ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS DEVPAD_AM_6M
    
    , FWD_AB
    , SPREAD_AB
    , CDI
    , FL_FIM_MES
    
    FROM 
    (SELECT DI_B.DATA as DT
    , to_char(ROUND(CAST(DI_A.VERTICE AS FLOAT)/252,2))  || 'x' || TO_CHAR(ROUND(CAST(DI_B.VERTICE AS FLOAT)/252,2)) || 'x' || TO_CHAR(ROUND(CAST(DI_C.VERTICE AS FLOAT)/252,2)) AS FLY
    , DI_A.VERTICE AS CONTRATO_A, DI_B.VERTICE AS CONTRATO_B, DI_C.VERTICE AS CONTRATO_C
    , DI_A.TAXA AS TAXA_A, DI_B.TAXA AS TAXA_B, DI_C.TAXA AS TAXA_C
    
    , ((1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)))-1 AS FWD_AB
    , ((1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)))-1 AS FWD_BC
    ,  (1+DI_C.TAXA)**(CAST(DI_C.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE))/(1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_C.VERTICE-DI_B.VERTICE)) - (1+DI_B.TAXA)**(CAST(DI_B.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE))/(1+DI_A.TAXA)**(CAST(DI_A.VERTICE AS FLOAT)/(DI_B.VERTICE-DI_A.VERTICE)) AS FLY_ABC
    ,  (DI_B.TAXA - DI_A.TAXA) as SPREAD_AB
    ,  (DI_C.TAXA - DI_B.TAXA) as SPREAD_BC
    ,  (DI_C.TAXA - DI_B.TAXA) - (DI_B.TAXA - DI_A.TAXA) as FLY_PONTOS
    
    , (CAST(DI_B.VERTICE AS FLOAT)/(1+DI_B.TAXA) - CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA))/(CAST(DI_A.VERTICE AS FLOAT)/(1+DI_A.TAXA)-CAST(DI_C.VERTICE AS FLOAT)/(1+DI_C.TAXA)) AS ALFA
    , ((IDX.VALOR/3000+1)**252.000000-1)*100 AS CDI
    , NVL(DU.FL_FIM_MES, DU.FL_D1) as FL_FIM_MES
    FROM DL_GESTAO_RECURSOS.CURVA DI_B
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_A ON DI_A.DATA = DI_B.DATA AND DI_A.VERTICE < DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.CURVA DI_C ON DI_C.DATA = DI_B.DATA AND DI_C.VERTICE > DI_B.VERTICE
    LEFT JOIN DL_GESTAO_RECURSOS.IDX_INDEX IDX ON IDX.DATA = DI_B.DATA AND IDX.CODINDEX = 'CDI'
    INNER JOIN dl_Gestao_recursos.vw_du DU ON DU.DT = DI_B.DATA AND ({fl_fim_mes_sql} = 1 or du.fl_d1 = 1)
    --WHERE 
    --DI_B.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_A.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520') and
    --DI_C.VERTICE IN ('126','189','252','315','378','441','504','567','630','756','882','1008','1134','1260','1512','1764','2016','2268','2520')
    ) VW
    WHERE VW.DT between '{data_inicial}' and '{data_final}'
    AND VW.FLY IN ('{lista_string}')
    """

   
    fly_df = pd.read_sql(Select, session) 
    fly_df.DT = pd.to_datetime(fly_df.DT, format='%Y-%m-%d')
    if fim_mes  == True:
        fly_df = fly_df[fly_df.FL_FIM_MES == 1]
   
    if 1 == 1:   
        fly_grafico = fly_df[fly_df.FLY == x] 
        fly_grafico = fly_grafico.sort_values(by=['DT'])
        fly_grafico = fly_grafico.reset_index() 
        fly_grafico = fly_grafico[fly_grafico.FL_FIM_MES == 1]
        pd.set_option('mode.chained_assignment', None)
    
        fig = make_subplots(specs=[[{"secondary_y": True}]])
        fig.add_vline(x=datetime(2014,11,15), line_width=3, line_dash="dash", line_color="green")
        TAXA = go.Scatter(x = fly_grafico.DT,y = fly_grafico['TAXA']*10000,mode = 'lines', name = 'Taxa', line = {'color': 'dodgerblue', 'width':1.5}, yaxis = 'y1')
        MEDIA = go.Scatter(x = fly_grafico.DT,y = fly_grafico['MEDIA_6M']*10000,mode = 'lines', name = 'Media', line = {'color': 'violet', 'width':1.5},yaxis = 'y1')
        LIM_INF = go.Scatter(x = fly_grafico.DT,y = fly_grafico['LIM_INF_6M']*10000,mode = 'lines', name = 'Liminte Inferior', line = {'color': 'violet','width':1},yaxis = 'y1')
        LIM_SUP = go.Scatter(x = fly_grafico.DT,y = fly_grafico['LIM_SUP_6M']*10000,mode = 'lines', name = 'Limite Superior', line = {'color': 'violet','width':1},yaxis = 'y1')
    
        CDI = go.Scatter(x = fly_grafico.DT,y = fly_grafico['CDI'],mode = 'lines', name = 'CDI', line = {'color': 'goldenrod', 'width':0.5,'dash':'dash'}, yaxis = 'y2')
      
        data  = [TAXA, MEDIA, LIM_SUP,LIM_INF, CDI]
        layout = dict(title = x,
                           xaxis = dict(title = 'Data')
                           ,yaxis = dict(title = 'Taxa')
                           #,xaxis_rangeslider_visible=True
                           ,yaxis2=dict(title='CDI',
                                       overlaying='y1',
                                       side='right')
                           )
        fig.update_layout(
            updatemenus=[
                dict(
                    buttons=[
                        dict(
                            label=FLY,
                            method="update",
                            args=[{"visible": [FLY == f for f in fly_df['FLY']]}],
                        )
                        for FLY in valores_fly
                    ],
                    direction="down",
                    pad={"r": 10, "t": 10},
                    showactive=True,
                    x=0.1,
                    xanchor="left",
                    y=1.1,
                    yanchor="top"
                ),
            ]
        )
            

        fig = dict(data=data, layout=layout)
        pio.show(fig)  
        end = time.time()
        print("Execution Time: %s secs." % "{:.2f}".format(end-start))
```
