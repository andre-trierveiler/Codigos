```python
##### Analise de Passivo ######
janela_movel = 20 #define a janela para a soma movel da captação/resgate
janela_movel_pl = 1 #define a janela para a soma movel da captação/resgate em percentual do PL

df_hist_classe_sub.data_competencia = pd.to_datetime(df_hist_classe_sub.data_competencia, format='ISO8601')
df_hist_classe_sub.data_atualizacao = pd.to_datetime(df_hist_classe_sub.data_atualizacao,  format='ISO8601')

#### Novos campospara analise ####

df_hist_classe_sub['ticket_medio'] = df_hist_classe_sub.valor_patrimonio_liquido/df_hist_classe_sub.numero_cotistas
df_hist_classe_sub['captacao_acumulada'] = df_hist_classe_sub.groupby('codigo_fundo').valor_volume_total_aplicacoes.rolling(555000, min_periods=1).sum().fillna(0).reset_index(0,drop=True)  
df_hist_classe_sub['resgate_acumulado'] = df_hist_classe_sub.groupby('codigo_fundo').valor_volume_total_resgates.rolling(555000, min_periods=1).sum().fillna(0).reset_index(0,drop=True)
df_hist_classe_sub['captacao_liquida_acumulada'] = df_hist_classe_sub['captacao_acumulada'] -df_hist_classe_sub['resgate_acumulado']  
df_hist_classe_sub['captacao_movel'] = df_hist_classe_sub.groupby('codigo_fundo').valor_volume_total_aplicacoes.rolling(janela_movel).sum().fillna(0).shift(1).reset_index(0,drop=True)  
df_hist_classe_sub['resgate_movel'] = df_hist_classe_sub.groupby('codigo_fundo').valor_volume_total_resgates.rolling(janela_movel).sum().shift(1).reset_index(0,drop=True)
df_hist_classe_sub['captacao_liquida_movel'] = df_hist_classe_sub['captacao_movel'] -df_hist_classe_sub['resgate_movel']  
df_hist_classe_sub['captacao_liquida'] = df_hist_classe_sub['valor_volume_total_aplicacoes'] -df_hist_classe_sub['valor_volume_total_resgates']
df_hist_classe_sub['captacao_pl'] = df_hist_classe_sub.valor_volume_total_aplicacoes/df_hist_classe_sub.valor_patrimonio_liquido
df_hist_classe_sub['resgate_pl'] = df_hist_classe_sub.valor_volume_total_resgates/df_hist_classe_sub.valor_patrimonio_liquido
df_hist_classe_sub['captacao_liquida_pl'] = df_hist_classe_sub.captacao_liquida/df_hist_classe_sub.valor_patrimonio_liquido
df_hist_classe_sub['captacao_pl_movel'] = df_hist_classe_sub.groupby('codigo_fundo').captacao_pl.rolling(janela_movel_pl).mean().fillna(0).reset_index(0,drop=True)
df_hist_classe_sub['resgate_pl_movel'] =  df_hist_classe_sub.groupby('codigo_fundo').resgate_pl.rolling(janela_movel_pl).mean().fillna(0).reset_index(0,drop=True)
df_hist_classe_sub['captacao_liquida_pl_movel'] = df_hist_classe_sub.groupby('codigo_fundo').captacao_liquida_pl.rolling(janela_movel_pl).mean().fillna(0).reset_index(0,drop=True)

patrimonio_inicial = df_hist_classe_sub.groupby('nome_comercial_classe').apply(lambda x: x.loc[x['data_competencia'].idxmin(), 'valor_patrimonio_liquido']).to_dict() 
df_hist_classe_sub['captacao_acumulada_pl'] = df_hist_classe_sub.apply( lambda row: row['captacao_acumulada'] / patrimonio_inicial[row['nome_comercial_classe']], axis=1 )
df_hist_classe_sub['resgate_acumuladp_pl'] = df_hist_classe_sub.apply( lambda row: row['resgate_acumulado'] / patrimonio_inicial[row['nome_comercial_classe']], axis=1 )
df_hist_classe_sub['captacao_liquida_acumulada_pl'] = df_hist_classe_sub['captacao_acumulada_pl'] -df_hist_classe_sub['resgate_acumuladp_pl']

num_cotistas_inicial = df_hist_classe_sub.groupby('nome_comercial_classe').apply(lambda x: x.loc[x['data_competencia'].idxmin(), 'numero_cotistas']).to_dict() 
df_hist_classe_sub['evolução_num_cotistas'] = df_hist_classe_sub.apply( lambda row: row['numero_cotistas'] / num_cotistas_inicial[row['nome_comercial_classe']], axis=1 )-1

ticket_medio_inicial = df_hist_classe_sub.groupby('nome_comercial_classe').apply(lambda x: x.loc[x['data_competencia'].idxmin(), 'ticket_medio']).to_dict() 
df_hist_classe_sub['evolução_ticket_medio'] = df_hist_classe_sub.apply( lambda row: row['ticket_medio'] / ticket_medio_inicial[row['nome_comercial_classe']], axis=1 )-1
```

```python
import numpy as np
from scipy.stats import linregress
import pandas as pd


def sharpe_ratio(returns, risk_free_rates):
    """
    Calcula o Índice de Sharpe para um investimento.

    Argumentos:
    returns (array_like): Vetor dos retornos diários do investimento.
    risk_free_rates (array_like): Vetor das taxas de retorno diário do ativo livre de risco.

    Retorna:
    float: Índice de Sharpe do investimento.
    """
    
    excess_returns = ((returns - risk_free_rates)+1)
    mean_excess_return = np.mean(excess_returns)
    std_excess_return = np.std(excess_returns)

    sharpe_ratio = mean_excess_return / std_excess_return

    return sharpe_ratio
```

```python
teste = sharpe_ratio(retorno, bench)
display(teste)
```

```python
def sortino_ratio(returns, risk_free_rates):
    """
    Calcula o Índice de Sortino para um investimento.

    Argumentos:
    returns (array_like): Vetor dos retornos diários do investimento.
    risk_free_rates (array_like): Vetor das taxas de retorno diário do ativo livre de risco.

    Retorna:
    float: Índice de Sortino do investimento.
    """
    
    
    excess_returns = returns - risk_free_rates
    #negative_returns = np.array([ret if ret < 0 else 0 for ret in returns])
    negative_returns = np.array([ret if ret < 0 else 0 for ret in excess_returns])
    mean_excess_return = np.mean(excess_returns)
    downside_deviation = np.std(negative_returns)

    sortino_ratio = mean_excess_return / downside_deviation

    return sortino_ratio
```

```python
teste = sortino_ratio(retorno, bench)
display(teste)
```

```python
def information_ratio(returns, benchmark_returns):
    """
    Calcula o Índice de Informação para um investimento em relação a um benchmark.

    Argumentos:
    returns (array_like): Vetor dos retornos diários do investimento.
    benchmark_returns (array_like): Vetor dos retornos diários do benchmark.

    Retorna:
    float: Índice de Informação do investimento em relação ao benchmark.
    """

    excess_returns = returns - benchmark_returns
    mean_excess_return = np.mean(excess_returns)
    tracking_error = np.std(excess_returns)

    information_ratio = mean_excess_return / tracking_error

    return information_ratio
```

```python
teste = information_ratio(retorno, bench)
display(teste)
```

```python


def treynor_ratio(returns, benchmark_returns):
    """
    Calcula o Índice de Treynor para um investimento em relação a um benchmark.

    Argumentos:
    returns (array_like): Vetor dos retornos diários do investimento.
    benchmark_returns (array_like): Vetor dos retornos diários do benchmark.

    Retorna:
    float: Índice de Treynor do investimento em relação ao benchmark.
    """

    excess_returns = returns - benchmark_returns
    beta, _, _, _, _ = linregress(benchmark_returns, excess_returns)

    treynor_ratio = np.mean(excess_returns) / beta

    return treynor_ratio
```

```python
def sharpe_ratio(returns, risk_free_rates):
    """
    Calcula o Índice de Sharpe para um investimento.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    risk_free_rates (pd.Series): Série das taxas de retorno diário do ativo livre de risco.

    Retorna:
    float: Índice de Sharpe do investimento.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(risk_free_rates.index)
    returns = returns.loc[common_dates]
    risk_free_rates = risk_free_rates.loc[common_dates]

    N = len(returns)

    excess_returns = (returns - risk_free_rates)
    mean_excess_return = (np.sum(excess_returns)+1)**(252/N)-1
    std_excess_return = np.std(excess_returns) * np.sqrt(252)

    sharpe_ratio = mean_excess_return / std_excess_return

    return sharpe_ratio
```

```python
def sortino_ratio2(returns, risk_free_rates):
    """
    Calcula o Índice de Sortino para um investimento.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    risk_free_rates (pd.Series): Série das taxas de retorno diário do ativo livre de risco.

    Retorna:
    float: Índice de Sortino do investimento.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(risk_free_rates.index)
    returns = returns.loc[common_dates]
    risk_free_rates = risk_free_rates.loc[common_dates]

    #negative_returns = returns.copy()
    #negative_returns[negative_returns >= 0] = 0

    excess_returns = returns - risk_free_rates
    negative_returns = np.array([ret if ret < 0 else 0 for ret in excess_returns])
    mean_excess_return = np.mean(excess_returns)
    downside_deviation = np.std(negative_returns)

    sortino_ratio = mean_excess_return / downside_deviation

    return sortino_ratio
```

```python
def beta(returns, benchmark_returns):
    """
    Calcula o coeficiente Beta para um investimento em relação a um benchmark.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.

    Retorna:
    float: Coeficiente Beta do investimento em relação ao benchmark.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(benchmark_returns.index)
    returns = returns.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    _, _, beta, _, _ = linregress(benchmark_returns, returns)

    return beta

def alpha(returns, benchmark_returns, risk_free_rate):
    """
    Calcula o Alfa para um investimento em relação a um benchmark.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    risk_free_rate (float): Taxa de retorno livre de risco.

    Retorna:
    float: Alfa do investimento em relação ao benchmark.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(benchmark_returns.index)
    returns = returns.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    excess_returns = returns - risk_free_rate - benchmark_returns
    alpha = np.mean(excess_returns)

    return alpha
```

```python
def tracking_error(returns, benchmark_returns):
    """
    Calcula o Tracking Error entre um investimento e um benchmark.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.

    Retorna:
    float: Tracking Error entre o investimento e o benchmark.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(benchmark_returns.index)
    returns = returns.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    excess_returns = returns - benchmark_returns
    tracking_error = np.std(excess_returns)

    return tracking_error

def sortino_ratio_modify(returns, risk_free_rates):
    """
    Calcula o Sortino Modificado para um investimento.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    risk_free_rates (pd.Series): Série das taxas de retorno diário do ativo livre de risco.

    Retorna:
    float: Sortino Modificado do investimento.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(risk_free_rates.index)
    returns = returns.loc[common_dates]
    risk_free_rates = risk_free_rates.loc[common_dates]

    negative_returns = returns.copy()
    negative_returns[negative_returns >= 0] = 0

    excess_returns = returns - risk_free_rates
    mean_excess_return = np.mean(excess_returns)
    downside_deviation = np.std(negative_returns)

    modified_sortino_ratio = mean_excess_return / downside_deviation

    return modified_sortino_ratio
```

```python
def sharpe_ratio_rolling(returns, risk_free_rates, window):
    """
    Calcula o Índice de Sharpe em uma janela móvel de acordo com o período especificado.

    Argumentos:
    returns (pd.Series): Série dos retornos diários do investimento.
    risk_free_rates (pd.Series): Série das taxas de retorno diário do ativo livre de risco.
    window (int): Período da janela móvel em dias.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Sharpe em janela móvel.
    """

    # Casamento de datas
    common_dates = returns.index.intersection(risk_free_rates.index)
    returns = returns.loc[common_dates]
    risk_free_rates = risk_free_rates.loc[common_dates]

    rolling_sharpe = []

    for i in range(len(returns) - window + 1):
        returns_window = returns.iloc[i:i+window]
        risk_free_rates_window = risk_free_rates.iloc[i:i+window]

        excess_returns = returns_window - risk_free_rates_window
        mean_excess_return = np.mean(excess_returns)
        std_excess_return = np.std(excess_returns)

        sharpe_ratio = mean_excess_return / std_excess_return
        rolling_sharpe.append(sharpe_ratio)

    index_dates = returns.index[window-1:]
    sharpe_df = pd.DataFrame(rolling_sharpe, index=index_dates, columns=["Sharpe Ratio"])

    return sharpe_df
```

```python
def indice_sharpe_movel(retornos, taxas_livres_de_risco, janela):
    """
    Calcula o Índice de Sharpe em uma janela móvel de acordo com o período especificado.

    Argumentos:
    retornos (pd.Series): Série dos retornos diários do investimento.
    taxas_livres_de_risco (pd.Series): Série das taxas de retorno diário do ativo livre de risco.
    janela (int): Período da janela móvel em dias.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Sharpe em janela móvel ou um índice global.
    """

    # Casamento de datas
    datas_comuns = retornos.index.intersection(taxas_livres_de_risco.index)
    retornos = retornos.loc[datas_comuns]
    taxas_livres_de_risco = taxas_livres_de_risco.loc[datas_comuns]

    if janela == 0:
        # Calcular o índice global para o período completo
        retornos_excedentes = retornos - taxas_livres_de_risco
        media_retorno_excedente = np.mean(retornos_excedentes)
        desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
        indice_sharpe = media_retorno_excedente / desvio_padrao_retorno_excedente
        sharpe_df = pd.DataFrame([indice_sharpe], index=[retornos.index[-1]], columns=["Índice Sharpe"])
    else:
        sharpe_movel = []
        for i in range(len(retornos)):
            # Ajustar a janela para os períodos iniciais
            if i < janela - 1:
                janela_retornos = retornos.iloc[:i+1]
                janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[:i+1]
            else:
                janela_retornos = retornos.iloc[i-janela+1:i+1]
                janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[i-janela+1:i+1]

            retornos_excedentes = janela_retornos - janela_taxas_livres_de_risco
            media_retorno_excedente = np.mean(retornos_excedentes)
            desvio_padrao_retorno_excedente = np.std(retornos_excedentes)

            indice_sharpe = media_retorno_excedente / desvio_padrao_retorno_excedente
            sharpe_movel.append(indice_sharpe)

        sharpe_df = pd.DataFrame(sharpe_movel, index=retornos.index, columns=["Índice Sharpe"])

    return sharpe_df
```

```python
################# PRONTO
def indice_sharpe_movel_multi(retornos_df, taxas_livres_de_risco=None, janela=0, janela_minima=1):
    """
    Calcula o Índice de Sharpe em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    taxas_livres_de_risco (pd.Series, opcional): Série das taxas de retorno diário do ativo livre de risco. Se não for fornecida, considera zero.
    janela (int): Período da janela móvel em dias.
    janela_minima (int): Número mínimo de dados para começar a calcular o Índice de Sharpe.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Sharpe em janela móvel para cada investimento.
    """

    # Casamento de datas
    if taxas_livres_de_risco is None:
        taxas_livres_de_risco = pd.Series(0, index=retornos_df.index)
    else:
        datas_comuns = retornos_df.index.intersection(taxas_livres_de_risco.index)
        retornos_df = retornos_df.loc[datas_comuns]
        taxas_livres_de_risco = taxas_livres_de_risco.loc[datas_comuns]

    sharpe_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_sharpe = []

        if janela == 0:
            # Calcular o índice global para o período completo
            retornos_excedentes = retornos_df[col] - taxas_livres_de_risco
            media_retorno_excedente = np.mean(retornos_excedentes)
            desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
            indice_sharpe = media_retorno_excedente / desvio_padrao_retorno_excedente
            col_sharpe.append(indice_sharpe)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[:i+1]
                    retornos_excedentes = janela_retornos - janela_taxas_livres_de_risco
                    media_retorno_excedente = np.mean(retornos_excedentes)
                    desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
                    indice_sharpe = media_retorno_excedente / desvio_padrao_retorno_excedente
                    col_sharpe.append(indice_sharpe)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[i-janela+1:i+1]
                    retornos_excedentes = janela_retornos - janela_taxas_livres_de_risco
                    media_retorno_excedente = np.mean(retornos_excedentes)
                    desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
                    indice_sharpe = media_retorno_excedente / desvio_padrao_retorno_excedente
                    col_sharpe.append(indice_sharpe)

            index_dates = retornos_df.index[janela_minima-1:]

        col_sharpe_df = pd.DataFrame(col_sharpe, index=index_dates, columns=[col])
        sharpe_movel = sharpe_movel.join(col_sharpe_df, how='outer')

    if janela == 0:
        return sharpe_movel.tail(1)
    
    return sharpe_movel.dropna()
```

```python
def information_ratio_rolling_multi(returns_df, benchmark_returns, window=0):
    """
    Calcula o Índice de Informação em uma janela móvel para múltiplos investimentos.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    window (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula o Índice de Informação para todo o período.
                             Se maior que 1, calcula o Índice de Informação em janela móvel. Padrão é 0.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Informação em janela móvel para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    if window <= 1:
        information_ratio_df = pd.DataFrame()

        for col in returns_df.columns:
            excess_returns = returns_df[col] - benchmark_returns
            mean_excess_return = np.mean(excess_returns)
            std_excess_return = np.std(excess_returns)

            information_ratio = mean_excess_return / std_excess_return
            information_ratio_df[col] = information_ratio

        return information_ratio_df

    else:
        rolling_information_ratio = {}

        for col in returns_df.columns:
            col_information_ratio = []

            for i in range(len(returns_df) - window + 1):
                returns_window = returns_df[col].iloc[i:i+window]
                benchmark_returns_window = benchmark_returns.iloc[i:i+window]

                excess_returns = returns_window - benchmark_returns_window
                mean_excess_return = np.mean(excess_returns)
                std_excess_return = np.std(excess_returns)

                information_ratio = mean_excess_return / std_excess_return
                col_information_ratio.append(information_ratio)

            index_dates = returns_df.index[window-1:]
            col_information_ratio_df = pd.DataFrame(col_information_ratio, index=index_dates, columns=[col])
            rolling_information_ratio[col] = col_information_ratio_df

        return pd.DataFrame(rolling_information_ratio)
```

```python
#######Pronto
def information_ratio_movel_multi(retornos_df, retornos_benchmark, janela=0, janela_minima=21):
    """
    Calcula o Índice de Informação em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    janela (int): Período da janela móvel em dias.
    janela_minima (int): Número mínimo de dados para começar a calcular o Índice de Informação.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Informação em janela móvel para cada investimento.
    """

    # Casamento de datas
    datas_comuns = retornos_df.index.intersection(retornos_benchmark.index)
    retornos_df = retornos_df.loc[datas_comuns]
    retornos_benchmark = retornos_benchmark.loc[datas_comuns]

    informacao_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_informacao = []

        if janela == 0:
            # Calcular o índice global para o período completo
            retornos_excedentes = retornos_df[col] - retornos_benchmark
            media_retorno_excedente = np.mean(retornos_excedentes)
            desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
            indice_informacao = media_retorno_excedente / desvio_padrao_retorno_excedente
            col_informacao.append(indice_informacao)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1]
                    retornos_excedentes = janela_retornos - janela_retornos_benchmark
                    media_retorno_excedente = np.mean(retornos_excedentes)
                    desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
                    indice_informacao = media_retorno_excedente / desvio_padrao_retorno_excedente
                    col_informacao.append(indice_informacao)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1]
                    retornos_excedentes = janela_retornos - janela_retornos_benchmark
                    media_retorno_excedente = np.mean(retornos_excedentes)
                    desvio_padrao_retorno_excedente = np.std(retornos_excedentes)
                    indice_informacao = media_retorno_excedente / desvio_padrao_retorno_excedente
                    col_informacao.append(indice_informacao)

            index_dates = retornos_df.index[janela_minima-1:]

        col_informacao_df = pd.DataFrame(col_informacao, index=index_dates, columns=[col])
        informacao_movel = informacao_movel.join(col_informacao_df, how='outer')

    if janela == 0:
        return informacao_movel.tail(1)
    
    return informacao_movel.dropna()
```

```python
def alpha_rolling_multi(returns_df, benchmark_returns, risk_free_rate, window=0):
    """
    Calcula o Alfa em uma janela móvel para múltiplos investimentos.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    risk_free_rate (float): Taxa de retorno livre de risco.
    window (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula o Alfa para todo o período.
                             Se maior que 1, calcula o Alfa em janela móvel. Padrão é 0.

    Retorna:
    pd.DataFrame: DataFrame contendo o Alfa em janela móvel para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    if window <= 1:
        alpha_df = pd.DataFrame()

        for col in returns_df.columns:
            excess_returns = returns_df[col] - risk_free_rate - benchmark_returns
            excess_returns_mean = excess_returns.mean()

            beta, alpha, _, _, _ = linregress(benchmark_returns, returns_df[col] - risk_free_rate)

            alpha_df[col] = [alpha]

        return alpha_df

    else:
        rolling_alpha = {}

        for col in returns_df.columns:
            col_alpha = []

            for i in range(len(returns_df) - window + 1):
                returns_window = returns_df[col].iloc[i:i+window]
                benchmark_returns_window = benchmark_returns.iloc[i:i+window]

                excess_returns = returns_window - risk_free_rate - benchmark_returns_window
                excess_returns_mean = excess_returns.mean()

                beta, alpha, _, _, _ = linregress(benchmark_returns_window, returns_window - risk_free_rate)

                col_alpha.append(alpha)

            index_dates = returns_df.index[window-1:]
            col_alpha_df = pd.DataFrame(col_alpha, index=index_dates, columns=[col])

            rolling_alpha[col] = col_alpha_df

        return pd.DataFrame(rolling_alpha)


def beta_rolling_multi(returns_df, benchmark_returns, window=0):
    """
    Calcula o Beta em uma janela móvel para múltiplos investimentos.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    window (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula o Beta para todo o período.
                             Se maior que 1, calcula o Beta em janela móvel. Padrão é 0.

    Retorna:
    pd.DataFrame: DataFrame contendo o Beta em janela móvel para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    if window <= 1:
        beta_df = pd.DataFrame()

        for col in returns_df.columns:
            beta_series = []

            for col in returns_df.columns:
                beta, _, _, _, _ = linregress(benchmark_returns, returns_df[col])

                beta_series.append(beta)

            beta_df[col] = beta_series

        return beta_df

    else:
        rolling_beta = {}

        for col in returns_df.columns:
            col_beta = []

            for i in range(len(returns_df) - window + 1):
                returns_window = returns_df[col].iloc[i:i+window]
                benchmark_returns_window = benchmark_returns.iloc[i:i+window]

                beta, _, _, _, _ = linregress(benchmark_returns_window, returns_window)

                col_beta.append(beta)

            index_dates = returns_df.index[window-1:]
            col_beta_df = pd.DataFrame(col_beta, index=index_dates, columns=[col])

            rolling_beta[col] = col_beta_df

        return pd.DataFrame(rolling_beta)
```

```python
def tracking_error_rolling_multi(returns_df, benchmark_returns, window=0):
    """
    Calcula o Tracking Error em uma janela móvel para múltiplos investimentos.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    window (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula o Tracking Error para todo o período.
                             Se maior que 1, calcula o Tracking Error em janela móvel. Padrão é 0.

    Retorna:
    pd.DataFrame: DataFrame contendo o Tracking Error em janela móvel para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    if window <= 1:
        tracking_error_df = pd.DataFrame()

        for col in returns_df.columns:
            tracking_error_series = np.std(returns_df[col] - benchmark_returns)

            tracking_error_df[col] = [tracking_error_series]

        return tracking_error_df

    else:
        rolling_tracking_error = {}

        for col in returns_df.columns:
            col_tracking_error = []

            for i in range(len(returns_df) - window + 1):
                returns_window = returns_df[col].iloc[i:i+window]
                benchmark_returns_window = benchmark_returns.iloc[i:i+window]

                tracking_error_series = np.std(returns_window - benchmark_returns_window)

                col_tracking_error.append(tracking_error_series)

            index_dates = returns_df.index[window-1:]
            col_tracking_error_df = pd.DataFrame(col_tracking_error, index=index_dates, columns=[col])

            rolling_tracking_error[col] = col_tracking_error_df

        return pd.DataFrame(rolling_tracking_error)


def treynor_ratio_rolling_multi(returns_df, benchmark_returns, risk_free_rate, window=0):
    """
    Calcula o Índice de Treynor em uma janela móvel para múltiplos investimentos.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    risk_free_rate (float): Taxa de retorno livre de risco.
    window (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula o Índice de Treynor para todo o período.
                             Se maior que 1, calcula o Índice de Treynor em janela móvel. Padrão é 0.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Treynor em janela móvel para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    if window <= 1:
        treynor_ratio_df = pd.DataFrame()

        for col in returns_df.columns:
            excess_returns = returns_df[col] - risk_free_rate
            excess_returns_mean = excess_returns.mean()

            beta, _, _, _, _ = linregress(benchmark_returns, returns_df[col])

            treynor_ratio_series = excess_returns_mean / beta

            treynor_ratio_df[col] = [treynor_ratio_series]

        return treynor_ratio_df

    else:
        rolling_treynor_ratio = {}

        for col in returns_df.columns:
            col_treynor_ratio = []

            for i in range(len(returns_df) - window + 1):
                returns_window = returns_df[col].iloc[i:i+window]
                benchmark_returns_window = benchmark_returns.iloc[i:i+window]

                excess_returns = returns_window - risk_free_rate
                excess_returns_mean = excess_returns.mean()

                beta, _, _, _, _ = linregress(benchmark_returns_window, returns_window)

                treynor_ratio_series = excess_returns_mean / beta

                col_treynor_ratio.append(treynor_ratio_series)

            index_dates = returns_df.index[window-1:]
            col_treynor_ratio_df = pd.DataFrame(col_treynor_ratio, index=index_dates, columns=[col])

            rolling_treynor_ratio[col] = col_treynor_ratio_df

        return pd.DataFrame(rolling_treynor_ratio)
```

```python
def calculate_investment_metrics(returns_df, benchmark_returns, risk_free_rate):
    """
    Calcula diversos índices de performance de investimentos para o período todo.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    risk_free_rate (float): Taxa de retorno livre de risco.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores dos índices para cada investimento.
    """

    # Casamento de datas
    common_dates = returns_df.index.intersection(benchmark_returns.index)
    returns_df = returns_df.loc[common_dates]
    benchmark_returns = benchmark_returns.loc[common_dates]

    metrics = {
        'Sharpe Ratio': [],
        'Treynor Ratio': [],
        'Information Ratio': [],
        'Tracking Error': [],
        'Beta': [],
        'Alpha': []
    }

    for col in returns_df.columns:
        excess_returns = returns_df[col] - risk_free_rate
        mean_excess_return = excess_returns.mean()
        std_excess_return = excess_returns.std()

        sharpe_ratio = mean_excess_return / std_excess_return

        beta, alpha, _, _, _ = linregress(benchmark_returns, returns_df[col])
        treynor_ratio = mean_excess_return / beta

        tracking_error = np.std(returns_df[col] - benchmark_returns)
        information_ratio = mean_excess_return / tracking_error

        metrics['Sharpe Ratio'].append(sharpe_ratio)
        metrics['Treynor Ratio'].append(treynor_ratio)
        metrics['Information Ratio'].append(information_ratio)
        metrics['Tracking Error'].append(tracking_error)
        metrics['Beta'].append(beta)
        metrics['Alpha'].append(alpha)

    metrics_df = pd.DataFrame(metrics, index=returns_df.columns).T

    return metrics_df

# Exemplo de uso:
# returns_df = pd.DataFrame({ ... })  # DataFrame com os retornos dos investimentos
# benchmark_returns = pd.Series([...])  # Série com os retornos do benchmark
# risk_free_rate = 0.01 / 252  # Taxa livre de risco diária

# metrics_df = calculate_investment_metrics(returns_df, benchmark_returns, risk_free_rate)
# print(metrics_df)
```

```python
def calculate_kappa(returns_df, benchmark_returns, p=1, risk_free_rate=0, window=None):
    """
    Calcula o índice de Kappa de ordem p para um DataFrame de retornos em comparação com um benchmark.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series): Série dos retornos diários do benchmark.
    p (int, opcional): Ordem do índice de Kappa. Padrão é 1.
    risk_free_rate (float ou pd.Series, opcional): Taxa de retorno livre de risco (fixa ou uma série de valores diários). Padrão é 0.
    window (int, opcional): Tamanho da janela móvel em dias. Padrão é None (calcula para todo o período).

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do índice de Kappa para cada investimento.
    """
    kappa_values = {}

    if window is None:
        common_dates = returns_df.index.intersection(benchmark_returns.index)
        if isinstance(risk_free_rate, pd.Series):
            common_dates = common_dates.intersection(risk_free_rate.index)

        returns_df = returns_df.loc[common_dates].dropna(how='any')
        benchmark_returns = benchmark_returns.loc[common_dates]
        if isinstance(risk_free_rate, pd.Series):
            risk_free_rate = risk_free_rate.loc[common_dates]

        for col in returns_df.columns:
            if isinstance(risk_free_rate, pd.Series):
                excess_returns = returns_df[col] - risk_free_rate
            else:
                excess_returns = returns_df[col] - risk_free_rate

            mean_excess_return = excess_returns.mean()
            l_moment_p = np.mean(np.abs(excess_returns) ** p) ** (1 / p)

            kappa = mean_excess_return / l_moment_p
            kappa_values[col] = kappa

    else:
        for i in range(len(returns_df) - window + 1):
            window_start = i
            window_end = i + window

            common_dates = returns_df.index[window_start:window_end].intersection(benchmark_returns.index)
            if isinstance(risk_free_rate, pd.Series):
                common_dates = common_dates.intersection(risk_free_rate.index)

            returns_window = returns_df.iloc[window_start:window_end].loc[common_dates].dropna(how='any')
            benchmark_returns_window = benchmark_returns.loc[common_dates]
            if isinstance(risk_free_rate, pd.Series):
                risk_free_rate_window = risk_free_rate.loc[common_dates]

            for col in returns_window.columns:
                if isinstance(risk_free_rate, pd.Series):
                    excess_returns = returns_window[col] - risk_free_rate_window
                else:
                    excess_returns = returns_window[col] - risk_free_rate

                mean_excess_return = excess_returns.mean()
                l_moment_p = np.mean(np.abs(excess_returns) ** p) ** (1 / p)

                kappa = mean_excess_return / l_moment_p
                if col not in kappa_values:
                    kappa_values[col] = []
                kappa_values[col].append(kappa)

    if window is None:
        kappa_df = pd.DataFrame.from_dict(kappa_values, orient='index', columns=[f'Kappa_{p}'])
    else:
        kappa_df = pd.DataFrame({f'Kappa_{p}': {col: np.mean(kappas) for col, kappas in kappa_values.items()}})

    return kappa_df
```

```python
#### Pronto #####

def calculate_kappa(retornos_df, retornos_benchmark, p=1, taxas_livres_de_risco=None, janela=0, janela_minima=1):
    """
    Calcula o índice de Kappa de ordem p para um DataFrame de retornos em comparação com um benchmark.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    p (int, opcional): Ordem do índice de Kappa. Padrão é 1.
    taxas_livres_de_risco (pd.Series, opcional): Série das taxas de retorno diário do ativo livre de risco. Se não for fornecida, considera zero.
    janela (int, opcional): Tamanho da janela móvel em dias. Se 0 ou 1, calcula o índice de Kappa para todo o período.
                             Se maior que 1, calcula o índice de Kappa em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular o índice de Kappa.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do índice de Kappa para cada investimento.
    """

    # Casamento de datas
    if taxas_livres_de_risco is None:
        taxas_livres_de_risco = pd.Series(0, index=retornos_df.index)
    else:
        datas_comuns = retornos_df.index.intersection(taxas_livres_de_risco.index)
        retornos_df = retornos_df.loc[datas_comuns]
        taxas_livres_de_risco = taxas_livres_de_risco.loc[datas_comuns]

    datas_comuns = retornos_df.index.intersection(retornos_benchmark.index)
    retornos_df = retornos_df.loc[datas_comuns]
    retornos_benchmark = retornos_benchmark.loc[datas_comuns]
    taxas_livres_de_risco = taxas_livres_de_risco.loc[datas_comuns]

    kappa_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_kappa = []

        if janela == 0:
            # Calcular o índice global para o período completo
            excess_returns = retornos_df[col] - taxas_livres_de_risco
            mean_excess_return = excess_returns.mean()
            l_moment_p = np.mean(np.abs(excess_returns) ** p) ** (1 / p)
            kappa = mean_excess_return / l_moment_p
            col_kappa.append(kappa)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1]
                    janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[:i+1]
                    excess_returns = janela_retornos - janela_taxas_livres_de_risco
                    mean_excess_return = excess_returns.mean()
                    l_moment_p = np.mean(np.abs(excess_returns) ** p) ** (1 / p)
                    kappa = mean_excess_return / l_moment_p
                    col_kappa.append(kappa)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1]
                    janela_taxas_livres_de_risco = taxas_livres_de_risco.iloc[i-janela+1:i+1]
                    excess_returns = janela_retornos - janela_taxas_livres_de_risco
                    mean_excess_return = excess_returns.mean()
                    l_moment_p = np.mean(np.abs(excess_returns) ** p) ** (1 / p)
                    kappa = mean_excess_return / l_moment_p
                    col_kappa.append(kappa)

            index_dates = retornos_df.index[janela_minima-1:]

        col_kappa_df = pd.DataFrame(col_kappa, index=index_dates, columns=[col])
        kappa_movel = kappa_movel.join(col_kappa_df, how='outer')

    if janela == 0:
        return kappa_movel.tail(1)
    
    return kappa_movel.dropna()
```

```python
### Pronto

def volatilidade_rolling_multi(retornos_df, retornos_benchmark, janela=0, janela_minima=1, agregador_retorno=1):
    """
    Calcula a volatilidade dos ativos em relação ao benchmark em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    janela (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula a volatilidade para todo o período.
                             Se maior que 1, calcula a volatilidade em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular a volatilidade.
    agregador_retorno (int, opcional): Número de dias para agregar os retornos antes de calcular a volatilidade. Padrão é 1 (retornos diários).

    Retorna:
    pd.DataFrame: DataFrame contendo a volatilidade em janela móvel para cada investimento.
    """

    # Casamento de datas
    datas_comuns = retornos_df.index.intersection(retornos_benchmark.index)
    retornos_df = retornos_df.loc[datas_comuns]
    retornos_benchmark = retornos_benchmark.loc[datas_comuns]

    # Agregar retornos se necessário
    if agregador_retorno > 1:
        retornos_df = retornos_df.rolling(agregador_retorno).sum().dropna()/agregador_retorno
        retornos_benchmark = retornos_benchmark.rolling(agregador_retorno).sum().dropna()/agregador_retorno

    volatilidade_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_volatilidade = []

        if janela == 0:
            # Calcular a volatilidade global para o período completo
            volatilidade = np.std(retornos_df[col] - retornos_benchmark)
            col_volatilidade.append(volatilidade)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1]
                    volatilidade = np.std(janela_retornos - janela_retornos_benchmark)
                    col_volatilidade.append(volatilidade)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1]
                    volatilidade = np.std(janela_retornos - janela_retornos_benchmark)
                    col_volatilidade.append(volatilidade)

            index_dates = retornos_df.index[janela_minima-1:]

        col_volatilidade_df = pd.DataFrame(col_volatilidade, index=index_dates, columns=[col])
        volatilidade_movel = volatilidade_movel.join(col_volatilidade_df, how='outer')

    if janela == 0:
        return volatilidade_movel.tail(1)
    
    return volatilidade_movel.dropna()
```

```python

##### Potno
def volatilidade_rolling_multi_compound(retornos_df, retornos_benchmark, janela=0, janela_minima=1, agregador_retorno=1):
    """
    Calcula a volatilidade dos ativos em relação ao benchmark em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    janela (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula a volatilidade para todo o período.
                             Se maior que 1, calcula a volatilidade em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular a volatilidade.
    agregador_retorno (int, opcional): Número de dias para agregar os retornos antes de calcular a volatilidade. Padrão é 1 (retornos diários).

    Retorna:
    pd.DataFrame: DataFrame contendo a volatilidade em janela móvel para cada investimento.
    """

    # Casamento de datas
    datas_comuns = retornos_df.index.intersection(retornos_benchmark.index)
    retornos_df = retornos_df.loc[datas_comuns]
    retornos_benchmark = retornos_benchmark.loc[datas_comuns]

    # Função para agregar retornos
    def agregar_retorno(x):
        return (x + 1).prod() - 1

    volatilidade_movel = pd.DataFrame(index=retornos_df.index)
    janela_minima = max(janela_minima, agregador_retorno)
    
    for col in retornos_df.columns:
        col_volatilidade = []

        if janela == 0:
            # Calcular a volatilidade global para o período completo
            volatilidade = np.std(retornos_df[col] - retornos_benchmark)
            col_volatilidade.append(volatilidade)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1].rolling(agregador_retorno).apply(agregar_retorno).dropna()
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1].rolling(agregador_retorno).apply(agregar_retorno).dropna()
                    if len(janela_retornos) > 0 and len(janela_retornos_benchmark) > 0:
                        volatilidade = np.std(janela_retornos - janela_retornos_benchmark)
                        col_volatilidade.append(volatilidade)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1].rolling(agregador_retorno).apply(agregar_retorno).dropna()
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1].rolling(agregador_retorno).apply(agregar_retorno).dropna()
                    if len(janela_retornos) > 0 and len(janela_retornos_benchmark) > 0:
                        volatilidade = np.std(janela_retornos - janela_retornos_benchmark)
                        col_volatilidade.append(volatilidade)
          
            index_dates = retornos_df.index[janela_minima-1:]

        col_volatilidade_df = pd.DataFrame(col_volatilidade, index=index_dates, columns=[col])
        volatilidade_movel = volatilidade_movel.join(col_volatilidade_df, how='outer')
    
    if janela == 0:
        return volatilidade_movel.tail(1)
    index_dates = retornos_df.index[janela_minima-1:]
    return volatilidade_movel.dropna()
    #return index_dates
```

```python

#### pronto
def correlacao_rolling_multi(retornos_df, retornos_benchmark, janela=0, janela_minima=1):
    """
    Calcula a correlação entre os retornos dos ativos e os retornos do benchmark em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    janela (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula a correlação para todo o período.
                             Se maior que 1, calcula a correlação em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular a correlação.

    Retorna:
    pd.DataFrame: DataFrame contendo a correlação em janela móvel para cada investimento.
    """

    # Casamento de datas
    datas_comuns = retornos_df.index.intersection(retornos_benchmark.index)
    retornos_df = retornos_df.loc[datas_comuns]
    retornos_benchmark = retornos_benchmark.loc[datas_comuns]

    correlacao_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_correlacao = []

        if janela == 0:
            # Calcular a correlação global para o período completo
            correlacao = retornos_df[col].corr(retornos_benchmark)
            col_correlacao.append(correlacao)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1]
                    correlacao = janela_retornos.corr(janela_retornos_benchmark)
                    col_correlacao.append(correlacao)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1]
                    correlacao = janela_retornos.corr(janela_retornos_benchmark)
                    col_correlacao.append(correlacao)

            index_dates = retornos_df.index[janela_minima-1:]

        col_correlacao_df = pd.DataFrame(col_correlacao, index=index_dates, columns=[col])
        correlacao_movel = correlacao_movel.join(col_correlacao_df, how='outer')

    if janela == 0:
        return correlacao_movel.tail(1)
    
    return correlacao_movel.dropna()
```

```python
def correlacao_media_rolling_multi(retornos_df, janela=0, janela_minima=1):
    """
    Calcula a correlação do retorno de cada um dos ativos contra o retorno médio dos demais ativos em uma janela móvel para múltiplos investimentos.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    janela (int, opcional): Período da janela móvel em dias. Se 0 ou 1, calcula a correlação para todo o período.
                             Se maior que 1, calcula a correlação em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular a correlação.

    Retorna:
    pd.DataFrame: DataFrame contendo a correlação em janela móvel para cada investimento.
    """

    correlacao_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_correlacao = []

        if janela == 0:
            # Calcular a correlação global para o período completo
            media_retorno_outros = retornos_df.drop(columns=[col]).mean(axis=1)
            correlacao = retornos_df[col].corr(media_retorno_outros)
            col_correlacao.append(correlacao)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df.iloc[:i+1]
                    media_retorno_outros = janela_retornos.drop(columns=[col]).mean(axis=1)
                    correlacao = janela_retornos[col].corr(media_retorno_outros)
                    col_correlacao.append(correlacao)
                else:
                    janela_retornos = retornos_df.iloc[i-janela+1:i+1]
                    media_retorno_outros = janela_retornos.drop(columns=[col]).mean(axis=1)
                    correlacao = janela_retornos[col].corr(media_retorno_outros)
                    col_correlacao.append(correlacao)

            index_dates = retornos_df.index[janela_minima-1:]

        col_correlacao_df = pd.DataFrame(col_correlacao, index=index_dates, columns=[col])
        correlacao_movel = correlacao_movel.join(col_correlacao_df, how='outer')

    if janela == 0:
        return correlacao_movel.tail(1)
    
    return correlacao_movel.dropna()
```

```python
def calculate_var(retornos_df, confidence_level=0.95, janela=0, janela_minima=1):
    """
    Calcula o Value at Risk (VaR) para um DataFrame de retornos em uma janela móvel.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    confidence_level (float, opcional): Nível de confiança para o cálculo do VaR. Padrão é 0.95 (95%).
    janela (int, opcional): Tamanho da janela móvel em dias. Se 0 ou 1, calcula o VaR para todo o período.
                             Se maior que 1, calcula o VaR em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular o VaR.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do VaR para cada investimento.
    """

    var_movel = pd.DataFrame(index=retornos_df.index)
    z_score = np.abs(np.percentile(np.random.normal(0, 1, 100000), (1 - confidence_level) * 100))

    for col in retornos_df.columns:
        col_var = []

        if janela == 0:
            # Calcular o VaR global para o período completo
            mean_return = retornos_df[col].mean()
            std_return = retornos_df[col].std()
            var = -(mean_return - z_score * std_return)
            col_var.append(var)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    mean_return = janela_retornos.mean()
                    std_return = janela_retornos.std()
                    var = -(mean_return - z_score * std_return)
                    col_var.append(var)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    mean_return = janela_retornos.mean()
                    std_return = janela_retornos.std()
                    var = -(mean_return - z_score * std_return)
                    col_var.append(var)

            index_dates = retornos_df.index[janela_minima-1:]

        col_var_df = pd.DataFrame(col_var, index=index_dates, columns=[col])
        var_movel = var_movel.join(col_var_df, how='outer')

    if janela == 0:
        return var_movel.tail(1)
    
    return var_movel.dropna().applymap(lambda x: -abs(x))
```

```python

def calculate_max_drawdown(retornos_df, retornos_benchmark):
    """
    Calcula o drawdown máximo para um DataFrame de retornos em comparação com um benchmark.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do drawdown máximo para cada investimento.
    """

    # Calcula os retornos acumulados
    cumulative_returns = (1 + retornos_df).cumprod()-1
    cumulative_benchmark = (1 + retornos_benchmark).cumprod()-1

    # Calcula o drawdown relativo ao benchmark
    relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0)
    drawdown = relative_returns.div(relative_returns.cummax(), axis=0)-1

    # Calcula o drawdown máximo
    max_drawdown = drawdown.min()

    # Retorna os resultados em um DataFrame com cada ativo em uma coluna
    max_drawdown_df = pd.DataFrame(max_drawdown, columns=['Max Drawdown']).T

    #return max_drawdown_df
    return max_drawdown_df
teste = calculate_max_drawdown(returns, bench)
display(teste)
```

```python
def calculate_average_drawdown(retornos_df, retornos_benchmark):
    """
    Calcula o drawdown médio para um DataFrame de retornos em comparação com um benchmark.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do drawdown médio para cada investimento.
    """

    # Calcula os retornos acumulados
    cumulative_returns = (1 + retornos_df).cumprod()-1
    cumulative_benchmark = (1 + retornos_benchmark).cumprod()-1

    # Calcula o drawdown relativo ao benchmark
    relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0)
    drawdown = relative_returns.div(relative_returns.cummax(), axis=0) - 1

    # Calcula o drawdown médio
    average_drawdown = drawdown.mean()

    # Retorna os resultados em um DataFrame com os nomes dos ativos nas colunas
    average_drawdown_df = pd.DataFrame(average_drawdown).T
    average_drawdown_df.columns = retornos_df.columns

    return average_drawdown_df
    
teste = calculate_average_drawdown(returns, bench)
display(teste)
```

```python
def calculate_drawdown_metrics(retornos_df, retornos_benchmark, janela=0, janela_minima=1):
    """
    Calcula o drawdown máximo e o drawdown médio para um DataFrame de retornos em comparação com um benchmark,
    utilizando janelas móveis e uma janela mínima.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.
    janela (int, opcional): Tamanho da janela móvel em dias. Se 0 ou 1, calcula os drawdowns para todo o período.
                             Se maior que 1, calcula os drawdowns em janela móvel. Padrão é 0.
    janela_minima (int): Número mínimo de dados para começar a calcular os drawdowns.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do drawdown máximo e do drawdown médio para cada investimento.
    """

    max_drawdown_movel = pd.DataFrame(index=retornos_df.index)
    average_drawdown_movel = pd.DataFrame(index=retornos_df.index)

    for col in retornos_df.columns:
        col_max_drawdown = []
        col_average_drawdown = []

        if janela == 0:
            # Calcular os drawdowns globais para o período completo
            cumulative_returns = (1 + retornos_df[col]).cumprod()
            cumulative_benchmark = (1 + retornos_benchmark).cumprod()
            relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0) - 1
            drawdown = relative_returns.div(relative_returns.cummax(), axis=0) - 1

            max_drawdown = drawdown.min()
            average_drawdown = drawdown.mean()

            col_max_drawdown.append(max_drawdown)
            col_average_drawdown.append(average_drawdown)
            index_dates = [retornos_df.index[-1]]
        else:
            for i in range(len(retornos_df)):
                # Ajustar a janela para os períodos iniciais
                if i < janela_minima - 1:
                    continue
                elif i < janela - 1:
                    janela_retornos = retornos_df[col].iloc[:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[:i+1]
                    cumulative_returns = (1 + janela_retornos).cumprod()
                    cumulative_benchmark = (1 + janela_retornos_benchmark).cumprod()
                    relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0) - 1
                    drawdown = relative_returns.div(relative_returns.cummax(), axis=0) - 1

                    max_drawdown = drawdown.min()
                    average_drawdown = drawdown.mean()

                    col_max_drawdown.append(max_drawdown)
                    col_average_drawdown.append(average_drawdown)
                else:
                    janela_retornos = retornos_df[col].iloc[i-janela+1:i+1]
                    janela_retornos_benchmark = retornos_benchmark.iloc[i-janela+1:i+1]
                    cumulative_returns = (1 + janela_retornos).cumprod()
                    cumulative_benchmark = (1 + janela_retornos_benchmark).cumprod()
                    relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0) - 1
                    drawdown = relative_returns.div(relative_returns.cummax(), axis=0) - 1

                    max_drawdown = drawdown.min()
                    average_drawdown = drawdown.mean()

                    col_max_drawdown.append(max_drawdown)
                    col_average_drawdown.append(average_drawdown)

            index_dates = retornos_df.index[janela_minima-1:]

        col_max_drawdown_df = pd.DataFrame(col_max_drawdown, index=index_dates, columns=[col])
        col_average_drawdown_df = pd.DataFrame(col_average_drawdown, index=index_dates, columns=[col])
        max_drawdown_movel = max_drawdown_movel.join(col_max_drawdown_df, how='outer')
        average_drawdown_movel = average_drawdown_movel.join(col_average_drawdown_df, how='outer')

    if janela == 0:
        return max_drawdown_movel.tail(1), average_drawdown_movel.tail(1)
    
    return max_drawdown_movel.dropna(), average_drawdown_movel.dropna()

# Exemplo de uso da função com dados fictícios


# Calcular os drawdowns máximo e médio com janelas móveis
max_dd_df, avg_dd_df = calculate_drawdown_metrics(returns, bench, janela=252, janela_minima=30)
display(max_dd_df)
display(avg_dd_df)
```

```python
def calculate_drawdown(retornos_df, retornos_benchmark):
    """
    Calcula o drawdown para um DataFrame de retornos em comparação com um benchmark.

    Argumentos:
    retornos_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    retornos_benchmark (pd.Series): Série dos retornos diários do benchmark.

    Retorna:
    pd.DataFrame: DataFrame contendo os valores do drawdown para cada investimento.
    """

    # Calcula os retornos acumulados
    cumulative_returns = (1 + retornos_df).cumprod()
    cumulative_benchmark = (1 + retornos_benchmark).cumprod()

    # Calcula o drawdown relativo ao benchmark
    relative_returns = cumulative_returns.div(cumulative_benchmark, axis=0) - 1
    drawdown = relative_returns.div(relative_returns.cummax(), axis=0) - 1

    return drawdown

# Calcular os drawdowns
drawdown_df = calculate_drawdown(returns, bench)

# Plotar o gráfico de linha do tempo do drawdown usando plotly express
drawdown_df.reset_index(inplace=True)
drawdown_long_df = drawdown_df.melt(id_vars='index', var_name='Ativo', value_name='Drawdown')

fig = px.line(drawdown_long_df, x='index', y='Drawdown', color='Ativo', title='Drawdowns Over Time')
fig.update_layout(xaxis_title='Date', yaxis_title='Drawdown', template='plotly_dark')

fig.show()
```

```python
def omega_index_rolling(returns_df, benchmark_returns=None, risk_free_rate=None, threshold=0, window=0, min_periods=1):
    """
    Calcula o Índice de Omega em uma janela móvel para cada ativo.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series, opcional): Série dos retornos diários do benchmark. Se None, usa retornos absolutos.
    risk_free_rate (pd.Series, opcional): Série da taxa livre de risco diária.
    threshold (float): Retorno limite para calcular o Índice de Omega. Padrão é 0.
    window (int): Período da janela móvel em dias. Se 0, calcula o Índice Omega para todo o período.
    min_periods (int): Número mínimo de dados necessários para começar o cálculo.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Omega em janela móvel para cada ativo.
    """
    omega_index_df = pd.DataFrame(index=returns_df.index)

    for col in returns_df.columns:
        rolling_omega = returns_df[col].rolling(window=window, min_periods=min_periods).apply(
            lambda x: (x[x > threshold].sum() / -x[x < threshold].sum()) if -x[x < threshold].sum() != 0 else np.inf,
            raw=True
        )
        omega_index_df[col] = rolling_omega

    return omega_index_df

teste = omega_index_rolling(returns, bench,risk_free_rate=None, threshold=0, window=5, min_periods=2)
display(teste)
```

```python
def calmar_index_rolling(returns_df, benchmark_returns=None, risk_free_rate=None, window=0, min_periods=1):
    """
    Calcula o Índice de Calmar em uma janela móvel para cada ativo.

    Argumentos:
    returns_df (pd.DataFrame): DataFrame com as datas como índice e os valores de retorno de n investimentos em cada coluna.
    benchmark_returns (pd.Series, opcional): Série dos retornos diários do benchmark. Se None, usa retornos absolutos.
    risk_free_rate (pd.Series, opcional): Série da taxa livre de risco diária. Não utilizado neste cálculo.
    window (int): Período da janela móvel em dias. Se 0, calcula o Índice Calmar para todo o período.
    min_periods (int): Número mínimo de dados necessários para começar o cálculo.

    Retorna:
    pd.DataFrame: DataFrame contendo o Índice de Calmar em janela móvel para cada ativo.
    """
    # Cálculo dos retornos acumulados
    cumulative_returns = (1 + returns_df).cumprod()

    if benchmark_returns is not None:
        cumulative_benchmark = (1 + benchmark_returns).cumprod()
        returns = cumulative_returns / cumulative_benchmark - 1
    else:
        returns = cumulative_returns

    # Cálculo do drawdown máximo
    max_drawdown = (returns / returns.cummax() - 1).rolling(window=window, min_periods=min_periods).min()
    
    # Retorno anualizado em janela móvel
    annualized_return = returns_df.rolling(window=window, min_periods=min_periods).mean() * 252

    calmar_index = annualized_return / abs(max_drawdown)
    return calmar_index



teste = calmar_index_rolling(returns, bench)# janela=252, janela_minima=30)
display(teste)
```
