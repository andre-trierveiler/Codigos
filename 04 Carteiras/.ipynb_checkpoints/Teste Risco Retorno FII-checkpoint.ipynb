```python

import numpy as np
import pandas as pd

from pandas_datareader import data as pdr
import yfinance as yf
yf.pdr_override()

from matplotlib import pyplot as plt
import plotly.express as px
import seaborn as sn

assets = ["RECR11.SA", "BTLG11.SA", "HCTR11.SA", "IRDM11.SA", "GSFI11.SA"]
## Define a data início
start = '2021-07-01'
## Define a data de fim
end = '2024-05-20'

# Busca os preços ajustados
prices = pdr.get_data_yahoo(assets, start = start, end = end)['Adj Close']

# Calcula os retornos e retira dados faltantes
returns = prices.pct_change().dropna()

# Calcula retorno mensal
returns.resample('M').agg(lambda x: (x + 1).prod()- 1)

# Calcula Retorno Acumulado
acum_index = (1 + returns).cumprod()

# Retira a data do index e transforma em coluna (para plotar)
acum = acum_index.reset_index(level = 0)

# Escolhe o tamanho do gráfico
plt.figure(figsize = (15, 10))

# Constrói o gráfico
fig = px.line(acum, x = 'Date', y = assets, title = 'Retorno Acumulado')
fig.show()





# Calcula Retorno Anualizado
N = len(acum_index)
annualized_return = acum_index.iloc[-1,] ** (12 / N) - 1

# Desvio padrão anualizado dos ativos
sd_annualized = returns.std() * np.sqrt(12)

# Cria um data frame de risco e retorno
annualized_risk_return = pd.DataFrame({'Risco Anualizado' : sd_annualized,
              'Retorno Anualizado' : annualized_return})

# Plota a relação Risco x Retorno
px.scatter(annualized_risk_return,
           x = 'Risco Anualizado',
           y = 'Retorno Anualizado',
           color =  annualized_risk_return.index,
           size = np.ones(len(returns.columns)),
           title = 'Risco x Retorno'
           )
```

```python
# Busca os preços ajustados
data = pdr.get_data_yahoo(['RECR11.SA', 'BTLG11.SA'], start = start, end = end)['Adj Close']

# Define portifolio de pesos iguais
def equal_weighted_portfolio(data):
   num_assets = len(data.columns)
   weights = np.ones(num_assets) / num_assets
   equal_weighted_returns = data.pct_change().dot(weights)
   equal_weighted_returns = pd.Series(equal_weighted_returns, name='s1')
   return equal_weighted_returns


# Portfolio de pesos por Market Cap
def market_cap_weighted_portfolio(data):
   market_caps = data.iloc[-1]  # Using latest market capitalisation
   weights = market_caps / market_caps.sum()
   market_cap_returns = data.pct_change().dot(weights)
   return market_cap_returns


# Portfolio de Minima Variência 
def minimum_variance_portfolio(data):
   cov_matrix = data.pct_change().cov()
   inv_cov_matrix = np.linalg.inv(cov_matrix)
   ones_vector = np.ones(len(data.columns))
   weights = inv_cov_matrix.dot(ones_vector) / ones_vector.dot(inv_cov_matrix).dot(ones_vector)
   minimum_var_returns = data.pct_change().dot(weights)
   return minimum_var_returns
```

```python
# Calculate portfolio returns using different strategies
equal_weighted_returns = equal_weighted_portfolio(data)
market_cap_returns = market_cap_weighted_portfolio(data)
minimum_var_returns = minimum_variance_portfolio(data)


# Visualise portfolio returns on separate graphs
plt.figure(figsize=(12, 12))


# Equal Weighted Portfolio
plt.subplot(3, 1, 1)
equal_weighted_returns.plot(label='Equal Weighted Portfolio')
plt.title('Equal Weighted Portfolio Returns')
plt.xlabel('Date')
plt.ylabel('Portfolio Return')
plt.legend()
plt.grid(True)


# Market Cap Weighted Portfolio
plt.subplot(3, 1, 2)
market_cap_returns.plot(label='Market Cap Weighted Portfolio')
plt.title('Market Cap Weighted Portfolio Returns')
plt.xlabel('Date')
plt.ylabel('Portfolio Return')
plt.legend()
plt.grid(True)


# Minimum Variance Portfolio
plt.subplot(3, 1, 3)
minimum_var_returns.plot(label='Minimum Variance Portfolio')
plt.title('Minimum Variance Portfolio Returns')
plt.xlabel('Date')
plt.ylabel('Portfolio Return')
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()
```

```python
# merge nos 3 portifolios

s1 = pd.Series(equal_weighted_returns, name='equal_weighted_returns')
s2 = pd.Series(market_cap_returns, name='market_cap_returns')
s3 = pd.Series(minimum_var_returns, name='minimum_var_returns') 

df = pd.concat([s1,s2,s3], axis=1, )


# Constrói o gráfico Retornos diários
fig = px.line(df, title = 'Retorno Diário')
fig.show()
```

```python
# Busca os preços ajustados
data = pdr.get_data_yahoo(['RECR11.SA', 'HCTR11.SA','GSFI11.SA'], start = start, end = end)['Adj Close']

# Retorno diários
returns = data.pct_change()

#Retorno Acumulado
returns_acum = (1 + data.pct_change()).cumprod() - 1


#Portfolio com maior sharpe
def maximum_sharpe_ratio_portfolio(returns):
   cov_matrix = returns.cov()
   inv_cov_matrix = np.linalg.inv(cov_matrix)
   ones_vector = np.ones(len(returns.columns))
   weights = inv_cov_matrix.dot(ones_vector) / ones_vector.dot(inv_cov_matrix).dot(ones_vector)
   return returns.dot(weights)


# Risk Parity Portfolio
def risk_parity_portfolio(returns):
   inv_cov_matrix = np.linalg.inv(returns.cov())
   weights = inv_cov_matrix.dot(np.ones(len(returns.columns))) / len(returns.columns)
   return returns.dot(weights)


# Inverse Volatility Weighted Portfolio
def inverse_volatility_weighted_portfolio(returns):
   weights = 1 / returns.std()
   weights /= weights.sum()
   return returns.dot(weights)

```

```python
# Busca os preços ajustados
assets =  ['RECR11.SA', 'BTLG11.SA', 'HCTR11.SA', 'IRDM11.SA', 'GSFI11.SA','XFIX11.SA','BOVA11.SA']
data = pdr.get_data_yahoo(assets, start = start, end = end)['Adj Close']

# Retorno diários
returns = data.pct_change()

#Retorno Acumulado
returns_acum = (1 + data.pct_change()).cumprod() - 1

# Constrói o gráfico Retornos diários
fig = px.line(returns, title = 'Retorno Diário')
fig.show()

# Constrói o gráfico Retornos diários
fig = px.line(returns_acum , title = 'Retorno Acumulado')
fig.show()


fig = px.imshow(returns.corr().round(decimals = 3), text_auto=True, title = 'Correlações')
fig.show()

#Portfolio com maior sharpe
def maximum_sharpe_ratio_portfolio(data, fl_weights = False):
   returns = data.pct_change()
   cov_matrix = returns.cov()
   inv_cov_matrix = np.linalg.inv(cov_matrix)
   ones_vector = np.ones(len(returns.columns))
   weights = inv_cov_matrix.dot(ones_vector) / ones_vector.dot(inv_cov_matrix).dot(ones_vector)
   if fl_weights == True:
       return weights
   else:  
       return returns.dot(weights)


# Risk Parity Portfolio
def risk_parity_portfolio(data, fl_weights = False):
   returns = data.pct_change() 
   inv_cov_matrix = np.linalg.inv(returns.cov())
   weights = inv_cov_matrix.dot(np.ones(len(returns.columns))) / len(returns.columns)
   if fl_weights == True:
       return weights
   else:      
       return returns.dot(weights)


# Inverse Volatility Weighted Portfolio
def inverse_volatility_weighted_portfolio(data, fl_weights = False):
   returns = data.pct_change() 
   weights = 1 / returns.std()
   weights = weights/weights.sum()
   if fl_weights == True:
       return weights
   else: 
        return returns.dot(weights)


# Define portifolio de pesos iguais
def equal_weighted_portfolio(data, fl_weights = False):
   returns = data.pct_change() 
   num_assets = len(returns.columns)
   weights = np.ones(num_assets) / num_assets
   equal_weighted_returns = returns.dot(weights)
   if fl_weights == True:
       return weights
   else:
       return equal_weighted_returns


# Portfolio de pesos por Market Cap
def market_cap_weighted_portfolio(data, fl_weights = False):
   returns = data.pct_change() 
   market_caps = data.iloc[-1]  # Using latest market capitalisation
   weights = market_caps / market_caps.sum()
   market_cap_returns = data.pct_change().dot(weights)
   if fl_weights == True:
       return weights
   else:  
       return market_cap_returns


# Portfolio de Minima Variência 
def minimum_variance_portfolio(data, fl_weights = False):
   returns = data.pct_change()
   cov_matrix = returns.cov()
   inv_cov_matrix = np.linalg.inv(cov_matrix)
   ones_vector = np.ones(len(returns.columns))
   weights = inv_cov_matrix.dot(ones_vector) / ones_vector.dot(inv_cov_matrix).dot(ones_vector)
   minimum_var_returns = returns.dot(weights)
   if fl_weights == True:
       return weights
   else: 
       return minimum_var_returns
    
    
```

```python
df_weights = pd.DataFrame({'maximum_sharpe_ratio_portfolio': maximum_sharpe_ratio_portfolio(data, fl_weights = True)
                 ,'risk_parity_portfolio': risk_parity_portfolio(data, fl_weights = True)
                 ,'equal_weighted_portfolio':equal_weighted_portfolio(data, fl_weights = True)
                 ,'minimum_variance_portfolio':minimum_variance_portfolio(data, fl_weights = True)

                 })

df_weights.index = assets

df_weights['inverse_volatility_weighted_portfolio'] = inverse_volatility_weighted_portfolio(data, fl_weights = True)
df_weights['market_cap_weighted_portfolio'] = market_cap_weighted_portfolio(data, fl_weights = True)

df_weights = df_weights.drop(columns = ['risk_parity_portfolio'])

#fig = px.bar(df_wights,  orientation='h')
#fig.show()

fig = px.bar(df_weights.transpose() ,  orientation='h', title = 'Pesos por Portifólio')
fig.show()
```
