```python
### Curva Vertices Fixos e Estatistica ####
Select  = f"""SELECT DATA as DT, TITULO
, VERTICE
, TAXA
, TAXA_DI
, PERC_CDI
, PERC_CDI_CORRENTE

, AVG(PERC_CDI) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_PERC_DI_PAPEL_VERTICE
, (STDDEV_SAMP(PERC_CDI) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as DVPAD_PERC_DI_PAPEL_VERTICE
--, (PERC_CDI - AVG(PERC_CDI) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))/((STDDEV_SAMP(PERC_CDI) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))) AS NUM_DVPAD_PERC_DI_PAPEL_VERTICE

, AVG(PERC_CDI) OVER (PARTITION BY VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_PERC_DI_VERTICE
,(STDDEV_SAMP(PERC_CDI) OVER (PARTITION BY VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as DVPAD_PERC_DIVERTICE
--, (PERC_CDI - AVG(PERC_CDI) OVER (PARTITION BY VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))/((STDDEV_SAMP(PERC_CDI) OVER (PARTITION BY VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))) AS NUM_DVPAD_PERC_DIVERTICE

, AVG(PERC_CDI_CORRENTE) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW) AS MEDIA_PERC_DI_CORRENTE_PAPEL_VERTICE
,(STDDEV_SAMP(PERC_CDI_CORRENTE) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW)) as DVPAD_PERC_DI_CORRENTE_PAPEL_VERTICE
--, (PERC_CDI_CORRENTE - AVG(PERC_CDI_CORRENTE) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))/((STDDEV_SAMP(PERC_CDI_CORRENTE) OVER (PARTITION BY TITULO,VERTICE ORDER BY DATA ROWS BETWEEN 125 PRECEDING AND CURRENT ROW))) AS NUM_DVPAD_PERC_DI_CORRENTE_PAPEL_VERTICE

, CUME_DIST() OVER (PARTITION BY TITULO, VERTICE ORDER BY PERC_CDI) AS CUME_PAPEL_VERTICE
, PERCENT_RANK() OVER (PARTITION BY TITULO, VERTICE ORDER BY PERC_CDI) AS RANK_PAPEL_VERTICE

, CUME_DIST() OVER (PARTITION BY  VERTICE ORDER BY PERC_CDI) AS CUME_DIST_VERTICE
, PERCENT_RANK() OVER (PARTITION BY VERTICE ORDER BY PERC_CDI) AS RANK_VERTICE

from(
SELECT TPF.DATA, TPF.TITULO, TPF.VERTICE, TPF.TAXA, DI.TAXA AS TAXA_DI
, CASE WHEN TITULO IN ('LTN') THEN ((1+TPF.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)/((1+ DI.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)*100 
  ELSE 
  CASE WHEN TITULO = 'LFT' THEN ((1+DI.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))*(1+TPF.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)/((1+DI.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)*100 
  ELSE 
  CASE WHEN TITULO IN ('NTN-F') THEN TPF.TAXA
  END END END AS PERC_CDI
, CASE WHEN TITULO = 'LFT' THEN ((1+idx.valor/3000)*(1+TPF.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)/(idx.valor/3000)*100 
  ELSE 
  CASE WHEN TITULO IN ('LTN') THEN ((1+TPF.TAXA)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)/((1+ idx.valor/3000)**(CAST(1 AS FLOAT)/CAST(252 AS FLOAT))-1)*100 
  ELSE NULL 
  END END AS PERC_CDI_CORRENTE

FROM DL_GESTAO_RECURSOS.CURVA_TPF tpf
left join dl_gestao_recursos.curva di on di.data = tpf.data and di.vertice = tpf.vertice
LEFT JOIN dl_gestao_recursos.idx_index idx on idx.data = tpf.data and codindex = 'CDI' /* corrente / 2y 10y*/
where tpf.data BETWEEN '{data_inicial_b}' AND '{data_final}'
and titulo in ('LFT','LTN','NTN-F','NTN-B') 
and tpf.VERTICE in ('42','63','126','252','504','630','1008','1659')
) tpf

--inner join dl_gestao_recursos.vw_du du on du.dt = tpf.data and du.fl_fim_mes = '1'
ORDER BY TITULO,TPF.VERTICE, data"""


df_a = pd.read_sql(Select, session) 
df_a.DT = pd.to_datetime(df_a.DT, format='%Y-%m-%d')
df_a = df_a.sort_values(by = ['TITULO', 'VERTICE','DT'])
```

```python
df_a = df_a.sort_values(by = ['TITULO', 'VERTICE','DT'])
```

```python
MENU_A = 'TITULO'
MENU_B = 'VERTICE'
df = df_a

options=[col for col in df.columns if col != "DT" and col !=MENU_A and col != MENU_B]
```

```python
MENU_A = 'TITULO'
MENU_B = 'VERTICE'
df = df_a
dropdown_a = widgets.Dropdown(
    options=df[MENU_A].unique(),
    description=MENU_A+':',
    disabled=False,
)

dropdown_b = widgets.Dropdown(
    options=df[MENU_B].unique(),
    description=MENU_B+':',
    disabled=False,
)

dropdown_valor = widgets.Dropdown(
    options=[col for col in df.columns if col != "DT" and col !=MENU_A and col != MENU_B],
    description='Valor Y:',
    disabled=False,
)

# Função para atualizar o gráfico com base na seleção do menu suspenso
def update_plot(menu_a, menu_b, valor_y):
    df_filtro = df[df[MENU_A] == dropdown_a.value]
    df_filtro = df_filtro[df_filtro[MENU_B] ==  dropdown_b.value]
    valor = valor_y.replace("_", "/")
    fig = px.line(df_filtro, x='DT', y=valor_y, color=MENU_B, title=f'Histórico - {menu_a} ({valor})')
    #fig = px.line(df_total, x='DtPos', y=valor_y)
    fig.show()
    

final = widgets.interactive_output(update_plot, {'menu_a': dropdown_a,'menu_b': dropdown_b, 'valor_y': dropdown_valor})

# Exibindo o menu suspenso e o gráfico inicial
display(dropdown_a, dropdown_b, dropdown_valor, final)
```
